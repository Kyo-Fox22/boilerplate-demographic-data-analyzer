```python
import pandas as pd
```

```python
# Read data from file
df = pd.read_csv('adult.data.csv')
df
```

```python
# Race Count Series with Race as index
df['race'].value_counts()
```

```python
# Average age of males
df[df['sex'] == 'Male']['age'].mean()
```

```python
# Percentage with Bachelors degrees
(df['education'] == 'Bachelors').sum()/len(df['education']) * 100
```

```python
# High/Advanced Education Percentage
high_educ = df['education'].str.contains(r'Bachelors|Masters|Doctorate')
round(high_educ.sum()/len(df) * 100, 2)
```

```python
# Low Education Percentage
low_educ = ~(df['education'].str.contains(r'Bachelors|Masters|Doctorate'))
round(low_educ.sum()/len(df) * 100, 2)
```

```python
# Percentage of high education rich
round((high_educ & (df['salary'] == '>50K')).sum()/len(df) * 100, 2)
```

```python
# Percentage of low education rich
round((low_educ & (df['salary'] == '>50K')).sum()/len(df) * 100, 2)
```

```python
# Minimum number of hours a person works per week
df['hours-per-week'].min()
```

```python
# Number of minimum hour workers
min_workers = (df['hours-per-week'] == df['hours-per-week'].min())
min_workers.sum()
```

```python
# Percentage of minimum hour workers that are rich
(min_workers & (df['salary'] == '>50K')).sum()/min_workers.sum() * 100
```

```python
# Country with highest percentage of rich
df[df['salary'] == '>50K']['native-country'].value_counts().index[0]
```

```python
# Percentage of rich people globally from the highest country (United-States)
df[df['salary'] == '>50K']['native-country'].value_counts()['United-States']/df[df['salary'] == '>50K']['native-country'].value_counts().sum() * 100
```

```python
# Top occupation for those who earn >50K in India
df[(df['salary'] == '>50K') & (df['native-country'] == 'India')]['occupation'].value_counts().index[0]
```
